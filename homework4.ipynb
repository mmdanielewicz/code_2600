## Question 1

#### Suppose an exam is to be given to students at a local college. To test its validity, the exam is administered to a sample of 10 students, and the scores are: 65, 70, 68, 73, 75, 80, 85, 78, 90, 100.

#### a. Calculate the sample mean x and standard deviation s of this set. To calculate the confidence interval [x − r, x + r], use the formula r = t * s/sqrt(n) where n is the number of records in the sample, and t = 2.262 (the critical value corresponding to the 95% confidence level and degrees of freedom n−1 = 9).

```python
scores = [65, 70, 68, 73, 75, 80, 85, 78, 90, 100]
n = 10
t = 2.262

mean =  np.mean(scores)
standard_dev = np.std(scores, ddof=1)
conf_int = t * (standard_dev / np.sqrt(n))

print(f"Mean: {mean}")
print(f"Standard Deviation: {standard_dev}")
print(f"95% Confidence Interval: +-{conf_int}")
```

#### b. Construct 40 bootstrapped samples of the data. (You might want to create a function in Python to do this quickly.) Calculate the mean of each sample, and use this set of means to estimate the 95% confidence interval for the mean of the population.

```python
b = 40
def boot_func(b, data):
    boot_means = []

    for i in range(b):
        rand_set = np.random.choice(data, size=10, replace=True)
        rand_mean = np.mean(rand_set)
        boot_means.append(rand_mean)

    return boot_means

boot_final = boot_func(b, scores)

lower = np.percentile(boot_final, 2.5)
upper = np.percentile(boot_final, 97.5)

print(f'Bootstrapped 95% Confidence Interval: between {lower} and {upper}')

```

#### c. Which method would you choose to estimate the confidence interval for this use case. Explain why you believe your chosen method is more reliable?

To estimate the confidence interval for this use case, I would choose the bootstrapping method. The dataset in this situations is small with only 10 students, and we can't be sure that the scores come from a normal distribution, which is an assumption that the normal t-based interval does. Bootstrapping lets us estimate the variation in the sample mean without relying on assumptions like this. If the sample size is small and the distribution isn't known, then bootstrapping is more reliable.
